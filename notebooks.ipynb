```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_excel("hok_makeup_dataset.xlsx")

# Show first rows
df.head()
```

```python
df.info()
df.columns
df.describe()
```

```python
df.info()

# Column names
df.columns

# Statistical summary
df.describe()
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df['City'] = df['City'].str.title()
df['Brand'] = df['Brand'].str.strip()
df['Channel'] = df['Channel'].str.strip()
```

```python
df['CTR'] = df['Clicks'] / df['Impressions']
```

```python
df['Conversion_Rate'] = df['Conversions'] / df['Clicks']
```

```python
df['CPC'] = df['Spend'] / df['Clicks']
```

```python
df['CPA'] = df['Spend'] / df['Conversions']
```

```python
df['ROAS'] = df['Revenue'] / df['Spend']
```

```python
df['Profit'] = df['Revenue'] - df['Spend']
```

```python
df['ROI'] = (df['Revenue'] - df['Spend']) / df['Spend']
```

```python
total_revenue = df['Revenue'].sum()
total_spend = df['Spend'].sum()
total_clicks = df['Clicks'].sum()
total_impressions = df['Impressions'].sum()
total_conversions = df['Conversions'].sum()

print("Total Revenue:", total_revenue)
print("Total Spend:", total_spend)
print("Total Clicks:", total_clicks)
print("Total Impressions:", total_impressions)
print("Total Conversions:", total_conversions)
```

```python
campaign_perf = df.groupby('Campaign Name')[['Revenue','Spend','Conversions']].sum()
campaign_perf.sort_values(by='Revenue', ascending=False)
```

```python
brand_perf = df.groupby('Brand')['Revenue'].sum().sort_values(ascending=False)
brand_perf
```

```python
channel_perf = df.groupby('Channel')[['Impressions','Clicks','Revenue']].sum()
channel_perf
```

```python
df.groupby('Brand')['Revenue'].sum().plot(kind='bar')
plt.title("Revenue by Brand")
plt.ylabel("Revenue")
plt.show()
```

```python
plt.scatter(df['Spend'], df['Revenue'])
plt.xlabel("Spend")
plt.ylabel("Revenue")
plt.title("Spend vs Revenue")
plt.show()
```

```python
df.groupby('Age Group')['Conversions'].sum().plot(kind='bar')
plt.title("Conversions by Age Group")
plt.show()
```

```python
df.groupby('Campaign Name')['ROI'].mean().sort_values(ascending=False)
```

```python
df.groupby('Channel')['ROAS'].mean().sort_values(ascending=False)
```

```python
df.groupby('Campaign Name')['Profit'].sum().sort_values(ascending=False)
```

```python
total_impressions = df['Impressions'].sum()
total_clicks = df['Clicks'].sum()
total_conversions = df['Conversions'].sum()

CTR = total_clicks / total_impressions
conversion_rate = total_conversions / total_clicks

print("Total Impressions:", total_impressions)
print("Total Clicks:", total_clicks)
print("Total Conversions:", total_conversions)

print("CTR:", CTR)
print("Conversion Rate:", conversion_rate)
```

```python
import matplotlib.pyplot as plt

funnel = [total_impressions, total_clicks, total_conversions]
labels = ['Impressions', 'Clicks', 'Conversions']

plt.bar(labels, funnel)
plt.title("Marketing Funnel")
plt.ylabel("Count")
plt.show()
```

```python
# Extract time features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Day_Name'] = df['Date'].dt.day_name()

df.head()
```

```python
monthly_revenue = df.groupby('Month')['Revenue'].sum()

monthly_revenue.plot(kind='bar')
plt.title("Monthly Revenue Performance")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
daily_revenue = df.groupby('Day_Name')['Revenue'].sum()

daily_revenue.plot(kind='bar')
plt.title("Revenue by Day of Week")
plt.xlabel("Day")
plt.ylabel("Revenue")
plt.show()
```

```python
daily_conversion = df.groupby('Day_Name')['Conversions'].sum()

daily_conversion.plot(kind='bar')
plt.title("Conversions by Day")
plt.show()
```

```python
df['Hour'] = pd.to_datetime(df['Time']).dt.hour
```

```python
hourly_revenue = df.groupby('Hour')['Revenue'].sum()

hourly_revenue.plot(kind='line')
plt.title("Revenue by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Revenue")
plt.show()
```

```python
monthly_roi = df.groupby('Month')['ROI'].mean()

monthly_roi.plot(kind='bar')
plt.title("Average ROI by Month")
plt.ylabel("ROI")
plt.show()
```

```python
time_analysis = df.groupby('Month').agg({
    'Revenue':'sum',
    'Spend':'sum',
    'Profit':'sum',
    'Conversions':'sum'
})

time_analysis
```

```python
df.groupby('Channel')['Revenue'].sum().sort_values(ascending=False).plot(kind='bar')

plt.title("Revenue by Marketing Channel")
plt.xlabel("Channel")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.show()
```

```python
df.groupby('Campaign Name')['ROI'].mean().sort_values().plot(kind='barh')

plt.title("ROI by Campaign")
plt.xlabel("ROI")
plt.ylabel("Campaign")

plt.show()
```
